Manipulação de strings e categorização: Dataset `Conversations (Português)`: Crie uma coluna `prompt_length` com o número de caracteres do texto do usuário, outra com o texto em caixa-alta (`str.upper()`). (https://huggingface.co/datasets/nicholasKluge/instruct-aira-datasetv2)

```python
# Instalar suporte para leitura de arquivos Parquet.
%pip install pyarrow
```

```python
# Carregar o dataset em portugues baixado do Hugging Face.
arquivo_parquet = 'data/instruct-aira-dataset-v2-portuguese.parquet'

df_huggingface = pd.read_parquet(arquivo_parquet)
df_huggingface.head()
```

```python
# Buscar o texto da primeira mensagem enviada pelo usuario.
def obter_texto_usuario(conversa):
    for mensagem in conversa:
        if mensagem.get('role') == 'user':
            return mensagem.get('content', '')
    return ''


# Criar as colunas pedidas a partir do texto do usuario.
df_huggingface['prompt_length'] = df_huggingface['conversations'].apply(
    lambda conversa: len(obter_texto_usuario(conversa))
)

df_huggingface['prompt_upper'] = df_huggingface['conversations'].apply(
    lambda conversa: obter_texto_usuario(conversa).upper()
)

df_huggingface[['prompt_length', 'prompt_upper']].head()
```

Aplicando funções com `apply()`: Dataset `prompts.csv`: Escreva uma função que classifica prompts como “curto” (<20 chars), “médio” ou “longo”, e aplique usando `df['prompt'].apply()`.

```python
# função para classificar o prompt como curto, médio ou longo
# curto: 0-19 caracteres
# médio: 20-470 caracteres
# longo: 471 ou mais caracteres
def classificar_prompt(length: str) -> str:
    if len(length)< 20:
        return 'curto'
    elif len(length) < 471:
        return 'médio'
    else:
        return 'longo'
# criar a coluna 'prompt_category' 
# em df usando a função classificar_prompt na coluna prompt
df['prompt_category']=df['prompt'].apply(lambda x: classificar_prompt(str(x)))
# incluir uma coluna com o tamanho do prompt
df['prompt_length'] = df['prompt'].apply(lambda x: len(str(x)))        
# exibir as primeiras linhas do dataframe
df.head()
```

Extraindo substrings com regex: Dataset `train.csv`: Use `df['text'].str.extract()` com expressões regulares para extrair menções a usuários (@username) e hashtags (#topic) dos tweets. (https://www.kaggle.com/datasets/abhi8923shriv/sentiment-analysis-dataset)

```python
%pip install charset-normalizer
```

```python

from charset_normalizer import from_path

resultado = from_path('train.csv').best()
print(resultado.encoding)
```

```python
import re
import pandas as pd

df_train = pd.read_csv('train.csv', encoding=resultado.encoding)

pattern_user = r'@(\w+)'
pattern_hashtag = r'#(\w+)'

df_train['user_mention'] = df_train['text'].str.extract(pattern_user)
df_train['hashtag'] = df_train['text'].str.extract(pattern_hashtag)

df_matches = df_train[df_train['user_mention'].notna() | df_train['hashtag'].notna()]

print('Menções encontradas:', df_train['user_mention'].notna().sum())
print('Hashtags encontradas:', df_train['hashtag'].notna().sum())

df_matches[['text', 'user_mention', 'hashtag']].head(10)
```

Junção entre tabelas (`merge`): Dataset `CRAG QA Pairs`: crie um DataFrame com perguntas e respostas a partir do arquivo JSONL compactado e junte com um DataFrame auxiliar de metadados de domínio.


```python
from pathlib import Path
import bz2
import json
from time import perf_counter

import pandas as pd

crag_url = 'https://github.com/facebookresearch/CRAG/raw/refs/heads/main/data/crag_task_1_and_2_dev_v4.jsonl.bz2?download='

jsonl_path = Path('crag_task_1_and_2_dev_v4.jsonl')
bz2_path = Path('crag_task_1_and_2_dev_v4.jsonl.bz2')

if bz2_path.exists():
    crag_path = bz2_path
    open_crag = lambda path: bz2.open(path, mode='rt', encoding='utf-8')
elif jsonl_path.exists():
    crag_path = jsonl_path
    open_crag = lambda path: open(path, mode='rt', encoding='utf-8')
else:
    raise FileNotFoundError(
        'Baixe o arquivo CRAG em crag_task_1_and_2_dev_v4.jsonl.bz2 ou '
        'crag_task_1_and_2_dev_v4.jsonl antes de executar esta celula.\n'
        f'URL: {crag_url}'
    )

print(f'Lendo {crag_path.name}...')
inicio = perf_counter()

columns = ['query', 'answer', 'domain', 'question_type', 'static_or_dynamic']
records = []
skipped_lines = 0

with open_crag(crag_path) as fp:
    for line_number, line in enumerate(fp, start=1):
        try:
            item = json.loads(line)
        except json.JSONDecodeError:
            skipped_lines += 1
            print(f'Linha {line_number} ignorada por JSON invalido.')
            continue

        records.append({column: item.get(column) for column in columns})
        print(f'Linha {line_number} processada.')

df_qa = pd.DataFrame.from_records(records).rename(columns={'query': 'question'})
tempo_total = perf_counter() - inicio

if skipped_lines:
    print(f'Linhas JSON invalidas ignoradas: {skipped_lines}')
```

```python

print(f'Registros carregados: {len(df_qa)} em {tempo_total:.1f}s')
print(df_qa['domain'].value_counts())

df_domain_metadata = pd.DataFrame(
    {
        'domain': ['finance', 'music', 'movie', 'open', 'sports'],
        'domain_description': [
            'Mercado financeiro e empresas.',
            'Artistas, albuns e faixas.',
            'Filmes, elenco e lancamentos.',
            'Conhecimento geral.',
            'Equipes, jogadores e estatisticas.'
        ],
        'source_type': [
            'economic_data',
            'entertainment_data',
            'entertainment_data',
            'general_web',
            'sports_data'
        ]
    }
)

df_crag_merged = df_qa.merge(df_domain_metadata, on='domain', how='left')
df_crag_merged['question'] = df_crag_merged['question'].str.slice(0, 70)

df_crag_merged[
    ['question', 'answer', 'domain', 'domain_description', 'source_type']
].head()
```
